```python
from PIL import Image
import matplotlib.pyplot as plt
img = Image.open("twin_towers.jpg")
print("Image Augmented functions: ")
print("Original Image")
plt.imshow(img)
plt.show()
print("Rotated Image")
rotated_img = img.rotate(45)
plt.imshow(rotated_img)
plt.show()
print("Transposed Image")
transposed_img = img.transpose(Image.FLIP_LEFT_RIGHT)
plt.imshow(transposed_img)
plt.show()
print("Resized Image")
resized_img = img.resize((50, 50))
plt.imshow(resized_img)
plt.show()
augmented_images = [img, rotated_img, transposed_img, resized_img]
titles = ["Original Image", "Rotated (45°)", "Transposed (FLIP_LEFT_RIGHT)", "Resized (100x100)"]
```
