```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'total_bill': [16.99, 10.34, 21.01, 23.68, 24.59, 15.69, 29.03, 12.69, 14.78, 10.27],
    'tip': [1.01, 1.66, 3.50, 3.31, 3.61, 3.12, 5.92, 2.00, 3.23, 1.71],
    'sex': ['Female', 'Male', 'Male', 'Male', 'Female', 'Male', 'Male', 'Male', 'Male', 'Male'],
    'smoker': ['No', 'No', 'No', 'No', 'No', 'No', 'No', 'No', 'No', 'No'],
    'day': ['Sun', 'Sun', 'Sun', 'Sun', 'Sun', 'Sun', 'Sat', 'Sat', 'Thur', 'Thur'],
    'time': ['Dinner', 'Dinner', 'Dinner', 'Dinner', 'Dinner', 'Dinner', 'Dinner', 'Dinner', 'Lunch', 'Lunch'],
    'size': [2, 3, 3, 2, 4, 4, 3, 2, 2, 2]
}
tips = pd.DataFrame(data)

print("=== Dataset Overview ===")
print(tips.head())

sns.displot(tips['total_bill'], kde=True)
sns.displot(tips['total_bill'], kde=False)

sns.jointplot(x='tip', y='total_bill', data=tips)
sns.jointplot(x='tip', y='total_bill', data=tips, kind="reg")
sns.jointplot(x='tip', y='total_bill', data=tips, kind="hex")

sns.pairplot(tips)
sns.pairplot(tips, hue='time')
sns.pairplot(tips, hue='day')

plt.figure()
sns.heatmap(tips.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")

plt.figure()
sns.boxplot(x=tips['total_bill'])
plt.title("Boxplot - Total Bill")

plt.figure()
sns.boxplot(x=tips['tip'])
plt.title("Boxplot - Tip")

print("\n=== Category Frequencies ===")
print(tips['time'].value_counts())

plt.figure()
sns.countplot(x='day', data=tips)

plt.figure()
sns.countplot(x='sex', data=tips)

plt.figure()
tips['sex'].value_counts().plot(kind='pie', autopct='%1.1f%%', title='Sex Distribution')

plt.figure()
tips['sex'].value_counts().plot(kind='bar', title='Sex Count Bar Chart')

plt.figure()
sns.countplot(x='day', data=tips[tips['time'] == 'Dinner'])
plt.title("Dinner Count by Day")

plt.show()
```
