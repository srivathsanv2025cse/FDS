```python
        #One Tail Z-Test

import numpy as np
import scipy.stats as stats

sample_data_1 = np.array([
    152, 148, 151, 149, 147, 153, 150, 148, 152, 149,
    151, 150, 149, 152, 151, 148, 150, 152, 149, 150,
    148, 153, 151, 150, 149, 152, 148, 151, 150, 153
])

population_mean = 150.0  # Hypothesized null value
sample_mean = np.mean(sample_data_1)
sample_std = np.std(sample_data_1, ddof=1)
n = len(sample_data_1)

# Z-statistic calculation
z_statistic = (sample_mean - population_mean) / (sample_std / np.sqrt(n))
p_value = 2 * (1 - stats.norm.cdf(np.abs(z_statistic)))

print(f"Sample Mean: {sample_mean:.2f}")
print(f"Z-Statistic: {z_statistic:.4f}")
print(f"P-Value: {p_value:.4f}")

alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: Mean is significantly different from 150 grams.")
else:
    print("Accept the null hypothesis: No significant difference from 150 grams.")


#Two Tail Z-Test
import numpy as np
from statsmodels.stats.weightstats import ztest

sample_data_1 = np.array([
    152, 148, 151, 149, 147, 153, 150, 148, 152, 149,
    151, 150, 149, 152, 151, 148, 150, 152, 149, 150,
    148, 153, 151, 150, 149, 152, 148, 151, 150, 153
])

sample_data_2 = np.array([
    155, 151, 154, 152, 150, 156, 153, 151, 155, 152,
    154, 153, 152, 155, 154, 151, 153, 155, 152, 153,
    151, 156, 154, 153, 152, 155, 151, 154, 153, 156
])

print(" ")

print(" ")
# Two-sample Z-test

z_stat, p_val = ztest(sample_data_1, sample_data_2)

print(f"Sample 1 Mean: {np.mean(sample_data_1):.2f}")
print(f"Sample 2 Mean: {np.mean(sample_data_2):.2f}")
print(f"Z-Statistic: {z_stat:.4f}")
print(f"P-Value: {p_val:.4e}")

alpha = 0.05
if p_val < alpha:
    print("Reject H0: There is a statistically significant difference between the two groups.")
else:
    print("Accept H0: No statistically significant difference between the two groups.")
```
