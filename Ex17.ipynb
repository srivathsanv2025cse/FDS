```python
import numpy as np
import pandas as pd
import pickle
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# Create DataFrame directly in code
data = {
    'YearsExperience': [
        1.1, 1.3, 1.5, 2.0, 2.2, 2.9, 3.0, 3.2, 3.2, 3.7,
        3.9, 4.0, 4.0, 4.1, 4.5, 4.9, 5.1, 5.3, 5.9, 6.0,
        6.8, 7.1, 7.9, 8.2, 8.7, 9.0, 9.5, 9.6, 10.3, 10.5
    ],
    'Salary': [
        39343, 46205, 37731, 43525, 39891, 56642, 60150, 54445, 64445, 57189,
        63218, 55794, 56957, 57081, 61111, 67938, 66029, 83088, 81363, 93940,
        91738, 98273, 101302, 113812, 109431, 105582, 116969, 112635, 122391, 121872
    ]
}

df = pd.DataFrame(data)

# Inspect and clean data
print(df.info())
df.dropna(inplace=True)
print(df.describe())

# Separate features (X) and target/label (y)
features = df.iloc[:, [0]].values  # YearsExperience
label = df.iloc[:, [1]].values     # Salary

# Split dataset into training and test sets (80% train, 20% test)
x_train, x_test, y_train, y_test = train_test_split(
    features, label, test_size=0.2, random_state=42
)

# Train Linear Regression model
model = LinearRegression()
model.fit(x_train, y_train)

# Model Evaluation
print("Train Score (R2):", model.score(x_train, y_train))
print("Test Score (R2):", model.score(x_test, y_test))
print("Coefficient (Slope):", model.coef_)
print("Intercept:", model.intercept_)

# Save trained model to disk
with open('SalaryPred.model', 'wb') as f:
    pickle.dump(model, f)

# Load model and make custom predictions
loaded_model = pickle.load(open('SalaryPred.model', 'rb'))

yr_of_exp = float(input("Enter Years of Experience: "))
yr_of_exp_NP = np.array([[yr_of_exp]])
estimated_salary = loaded_model.predict(yr_of_exp_NP)

print(f"Estimated Salary for {yr_of_exp} years of experience is {estimated_salary[0][0]:.2f}")
```
