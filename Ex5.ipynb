```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data.csv")

print("Data Exploration: ")
df.info()

display(df.describe())

print("Histogram: \n")
df.hist(bins=50, figsize=(20,15))
plt.show()

print("Pair Plot: \n")
sns.pairplot(df)
plt.show()
```
