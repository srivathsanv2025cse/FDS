```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
n = 100
sales = np.random.normal(50, 15, n).round(1)
df = pd.DataFrame({
    'Month': np.random.randint(1, 13, n),
    'Sales': sales,
    'Ad_Budget': sales * 0.7 + np.random.normal(5, 3, n),
    'Category': np.random.choice(['Electronics', 'Apparel', 'Home'], n, p=[0.4, 0.35, 0.25]),
    'Heights': np.random.normal(170, 5, n)
})

fig, ax = plt.subplots(2, 3, figsize=(14, 8))
df.groupby('Month')['Sales'].mean().plot(ax=ax[0, 0], marker='o', title='Line Plot (Trends)')
df.groupby('Category')['Sales'].sum().plot.bar(ax=ax[0, 1], color='green', title='Bar Chart (Comparison)')
ax[0, 2].scatter(df['Ad_Budget'], df['Sales'], c='red', alpha=0.6)
ax[0, 2].set_title('Scatter Plot (Correlation)')
sns.histplot(df['Heights'], kde=True, color='purple', ax=ax[1, 0]).set_title('Histplot (Distribution)')
sns.countplot(data=df, x='Category', hue='Category', palette='Set2', ax=ax[1, 1]).set_title('Countplot (Frequency)')
ax[1, 2].axis('off')
plt.tight_layout()
plt.show()

sns.pairplot(df[['Sales', 'Ad_Budget', 'Heights', 'Category']], hue='Category', palette='Set2')
plt.show()
```
