```python
import pandas as pd
import matplotlib as plt
import seaborn as sns
data = {
    'Year': [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024],
    'Job_Postings': [1200, 1500, 1900, 2500, 3200, 2800, 4200, 5100, 6200, 7500]
}
df = pd.DataFrame(data)
print("Data Science Job Postings:")
print(df)
print("\nStatistical Summary:")
print(df.describe())
df['Percentage_Change'] = df['Job_Postings'].pct_change() * 100
print("\nYear-on-Year Percentage Change:")
print(df)
plt.figure(figsize=(10, 6))

sns.lineplot(
    data=df,
    x='Year',
    y='Job_Postings',
    marker='o'
)

plt.title('Trend of Data Science Job Postings Over the Last Decade')
plt.xlabel('Year')
plt.ylabel('Number of Job Postings')
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
from pandas.core.frame import DataFrame
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = {
    'Job_Role' : ["Data Scientist",
"Data Analyst",
"Data Engineer",
"Data Scientist",
"Machine Learning Engineer",
"Data Analyst",
"Data Engineer",
"Data Scientist",
"Data Analyst",
"Business Analyst"]
}

df = pd.DataFrame(data)
print("Dataset:")
print(df.head())
print("\nColumns in dataset:")
print(df.columns)
role_counts = df['Job_Role'].value_counts()
print("\nDistribution of Data Science Roles:")
print(role_counts)
plt.figure(figsize=(10, 6))

sns.barplot(
    x=role_counts.index,
    y=role_counts.values
)

plt.title("Distribution of Data Science Roles")
plt.xlabel("Data Science Role")
plt.ylabel("Number of Job Postings")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
