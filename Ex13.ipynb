```python
import pandas as pd
import numpy as np

raw_data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 108, 108],
    'Age': [25, -5, 31, np.nan, 45, 150, 29, 38, 38],
    'Salary': [50000, 60000, np.nan, 55000, 120000, 62000, '$70,000', 58000, 58000],
    'Department': ['IT', 'HR', 'Finance', 'IT', 'Unknown_Dept', np.nan, 'HR', 'IT', 'IT'],
    'Experience_Yrs': [2, 5, 7, 3, 20, -1, 4, 10, 10]
}

df = pd.DataFrame(raw_data)

print("=== ORIGINAL DATASET ===")
print(df)
print("\n=== MISSING VALUES COUNT ===")
print(df.isnull().sum())

df = df.drop_duplicates()

df['Salary'] = df['Salary'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False)
df['Salary'] = pd.to_numeric(df['Salary'], errors='coerce')

df.loc[(df['Age'] < 18) | (df['Age'] > 70), 'Age'] = np.nan
df.loc[df['Experience_Yrs'] < 0, 'Experience_Yrs'] = np.nan
df['Department'] = df['Department'].replace('Unknown_Dept', np.nan)

df['Age'] = df['Age'].fillna(df['Age'].median())
df['Experience_Yrs'] = df['Experience_Yrs'].fillna(df['Experience_Yrs'].median())
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])

print("\n=== CLEANED DATASET ===")
print(df)
```
